```python
df['Ram'] = df['Ram'].str.replace('GB','').astype(int)
```

```python
df['Weight'] = df['Weight'].str.replace('kg','').astype(float)
```

```python
category = ['Company', 'TypeName', 'ScreenResolution', 'Cpu','Memory','Gpu','OpSys']
```

```python
df_encoded = pd.get_dummies(df, columns = category,drop_first = True)
```

```python
df_encoded
```

```python
df_encoded.select_dtypes(include='object').columns
```

```python
df_encoded.dtypes[df_encoded.dtypes == 'object']
```

```python
df_encoded.drop(columns=['Product'], inplace=True)
```

```python
df_encoded = df_encoded.astype(int)
```

```python
df_encoded.info()
```

```python
df_encoded.shape
```

```python
df_encoded.columns
```

```python
df_encoded.drop(columns=['laptop_ID'], inplace=True)
```

```python
df_encoded.shape
```

```python
df_encoded.info()
```

```python
from sklearn.preprocessing import StandardScaler
numerical_features = ['Inches','Ram','Weight']
scaler = StandardScaler()
df_encoded[numerical_features] = scaler.fit_transform(df_encoded[numerical_features])
```

```python
from sklearn.model_selection import train_test_split
X = df_encoded.drop(columns = ['Price_euros'])
y = df_encoded['Price_euros']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)
```

```python
X_train.shape
```

```python
y_train.shape
```

```python
#Building an ANN
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense,Dropout
from tensorflow.keras.callbacks import EarlyStopping
```

```python
model = Sequential()
```

```python
input_dim = X_train.shape[1]

model.add(Dense(units = 128, activation= 'relu', input_shape = (input_dim,)))
model.add(Dropout(0.2)) # 1 hidden layer

model.add(Dense(units = 128, activation= 'relu'))
model.add(Dropout(0.2)) # 2 hidden layer

model.add(Dense(units = 64, activation= 'relu'))
model.add(Dropout(0.2)) # 3 hidden layer

model.add(Dense(units = 32, activation= 'relu'))
model.add(Dropout(0.2))

```

```python
model.add(Dense(units = 1, activation='linear'))
```

```python
model.summary()
```

```python
early = EarlyStopping(monitor='val_loss',verbose=1,patience=5)
```

```python
model.compile(optimizer='adam',loss='mean_squared_error',metrics=['mean_absolute_error'])
```

```python
history = model.fit(X_train,y_train,epochs=100,batch_size=32,validation_split=0.2,verbose=1,callbacks=[early])
```

```python
y_pred = model.predict(X_test)
```

```python
y_pred_train = model.predict(X_train)
```

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
```

```python
mse = mean_squared_error(y_test,y_pred)
mae = mean_absolute_error(y_test,y_pred)
r2 = r2_score(y_test,y_pred)

print(mse)
print(mae)
print(r2)
```

```python
plt.plot(history.history['loss'], label='Training Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
```

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error,r2_score

#Training - Evaluation
mse = mean_squared_error(y_train, model.predict(X_train))
mae = mean_absolute_error(y_train, model.predict(X_train))
r2_score = r2_score(y_train, model.predict(X_train))

print(f"Training Mean Squared Error (MSE): {mse}")
print(f"Training Mean Absolute Error (MAE): {mae}")
print(f"Training R2 Score: {r2_score}")
```
